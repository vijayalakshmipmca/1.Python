```python
# Class and function
#age_calculaiton using Class and function : Non parametrized : with return 
class agecalculation():
    def agecalc():
        age = int(input("Enter your Age : "))
        if age < 0 :
            print("Invalid age")
            message = "Invalid age"
        elif (age >0 and age < 18):
            print("Minor")
            message = "Minor"
        elif age < 35:
            print("Adult")
            message  = "Adult"
        elif age < 60:
            print("Middle-aged")
            message = "Middle-aged"
        else:
            print("Senior Citizen")
            message = "Senior Citizen"
        return message
```

```python
#classname.functionname() to execute
agecalculation.agecalc()
```

```python
#Class and Function
#Odd Even number check : using Class and Function : Non parametrized : with return
class OddEvenCheck():
    def OddEven() :
        num = int(input("Enter a number: "))
        if num % 2 == 1:
            print("Odd Number")
            msg = "Odd Number"
        else:
            print("Even Number")
            msg = "Even Number"
        return msg
```

```python
message1= OddEvenCheck.OddEven()
if message1 == "Even Number" :
    print ("The given Number is Even Number")
else :
    print ("The given Number is Odd Number")
```

```python
#Class and Function 
#BMI Calculation
class BMICalculation():
    def bmicalc():
        bmi = int(input("Enter the BMI Index:"))
        if bmi <= 18 :
            print("Underweight")
            msg = "Underweight"
        elif bmi <= 25 :
            print("Normal")
            msg = "Normal"
        elif bmi <= 40 :
            print("Very Overweight")
            msg = "Very Overweight"
        else :
            print("Obese")
            msg = "Obese"
        return msg
```

```python
BMI = BMICalculation.bmicalc()
print ( "The Category for the BMI is : ", BMI)
```

```python
#Class and function
class MathCaculation():
#addtion : with parameters
    def addition (num1, num2) :
        add = num1+num2
        return add
#subtraction : with parameters        
    def subtraction (num1, num2) :
        sub = num1-num2
        return sub  
#multiplication : with parameters        
    def Multiple (num1, num2) :
        mul = num1 * num2
        return mul           
```

```python
add = MathCaculation.addition (16, 22)
print ("The Answer is : ", add)
```

```python
sub = MathCaculation.subtraction (22,10)
print ("The Answer is : ", sub)
```

```python
mul = MathCaculation.Multiple (22,10)
print ("The Answer is : ", mul)
```
