```python
#for loop
Lists = [23, 35, 10 ,0, 55, 60, -2, 200, 5,44]
for age in Lists:
if age < 0 :
    print(age, "InValid age")
elif (age >0 and age < 18):
    print(age, "Minor")
elif age < 35:
    print(age, "Adult")
elif age < 60:
    print(age, "Middle-aged")
else:
    print(age, "Senior Citizen")
```

```python
#for loop
Lists = [23, 35, 10 ,0, 55, 60, -2, 200, 5,44]
for age in Lists:
    if age <= 0 :
        print(age, "InValid age")
    elif (age >0 and age < 18):
        print(age, "Minor")
    elif age < 35:
        print(age, "Adult")
    elif age < 60:
        print(age, "Middle-aged")
    else:
        print(age, "Senior Citizen")
```
