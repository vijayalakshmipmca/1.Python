```python
#function
#age_calculaiton using function : Non parametrized : with return 
def agecalc():
    age = int(input("Enter your Age : "))
    if age < 0 :
        print("Invalid age")
        message = "Invalid age"
    elif (age >0 and age < 18):
        print("Minor")
        message = "Minor"
    elif age < 35:
        print("Adult")
        message  = "Adult"
    elif age < 60:
        print("Middle-aged")
        message = "Middle-aged"
    else:
        print("Senior Citizen")
        message = "Senior Citizen"
    return message
```

```python
Age_Category = agecalc()
print (Age_Category)
```

```python
#Function
#Odd Even number check : using function : Non parametrized : with return
def OddEven() :
    num = int(input("Enter a number: "))
    if num % 2 == 1:
        print("Odd Number")
        msg = "Odd Number"
    else:
        print("Even Number")
        msg = "Even Number"
    return msg
```

```python
message1= OddEven()
if message1 == "Even Number" :
    print ("The given Number is Even Number")
else :
    print ("The given Number is Odd Number")
```

```python
#Function 
#BMI Calculation
def bmicalc():
    bmi = int(input("Enter the BMI Index:"))
    if bmi <= 18 :
        print("Underweight")
        msg = "Underweight"
    elif bmi <= 25 :
        print("Normal")
        msg = "Normal"
    elif bmi <= 40 :
        print("Very Overweight")
        msg = "Very Overweight"
    else :
        print("Obese")
        msg = "Obese"
    return msg
```

```python
BMI = bmicalc()
print ( "The Category for the BMI is : ", BMI)
```
