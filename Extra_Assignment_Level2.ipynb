```python
# print Odd Numbers in the list
Num_list = (20,10,16,19,25,1,276,188)
print (Num_list)
for odd_num in Num_list:
    if (odd_num % 2) == 1 :
        print (odd_num, "is odd")
```

```python
# print Even numbers in the list
Num_list = (20,10,16,19,25,1,276,188)
print (Num_list)
for even_num in Num_list:
    if (even_num % 2) != 1 :
        print (even_num, "is even")
```
